```python
import numpy as np
import pandas as pd
import os
from IPython.display import display

path = "/host/data/mint/Motion_Dataset/Mixamo/prep/rdy_to_wan/5_frames/all/all_metadata_front-view_5frames.csv"
metadata = pd.read_csv(path)
motion_name = []
for i in range(len(metadata)):
    char, motion, cam, misc, chunk_id, ext = metadata.iloc[i]["video"].split("_")
    if motion in motion_name:
        continue
    motion_name.append(motion)
print(len(motion_name))

# random motion for train/test
np.random.seed(47)
motion_name = np.array(motion_name)
train_motions = np.random.choice(motion_name, size=100, replace=False)
test_motions = np.random.choice(motion_name, size=25, replace=False)

# Re-create new_csv file contains only {train, test} motions
new_train_csv = []
new_test_csv = []

for i in range(len(metadata)):
    char, motion, cam, misc, chunk_id, ext = metadata.iloc[i]["video"].split("_")
    if motion in train_motions:
        new_train_csv.append(metadata.iloc[i])
    elif motion in test_motions:
        new_test_csv.append(metadata.iloc[i])

new_train_csv = pd.DataFrame(new_train_csv, columns=metadata.columns)
new_test_csv = pd.DataFrame(new_test_csv, columns=metadata.columns)
os.makedirs("metadata_csv", exist_ok=True)
new_train_csv.to_csv("./metadata_csv/train_metadata_front-view_5frames_100motions.csv", index=False)
new_test_csv.to_csv("./metadata_csv/test_metadata_front-view_5frames_25motions.csv", index=False)
display(new_train_csv.head(5))
```
